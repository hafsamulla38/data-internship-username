```python
#Step 1: Import libraries

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set the chart style
sns.set_theme(style="whitegrid")
print("All libraries imported successfully!")
```

```python
#Step 2 — Prepare and Inspect the Retail Dataset.

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Category": [
        "Electronics", "Clothing", "Furniture", "Electronics",
        "Clothing", "Furniture", "Electronics", "Clothing",
        "Furniture", "Electronics", "Clothing", "Furniture",
        "Electronics", "Clothing", "Furniture", "Electronics",
        "Clothing", "Furniture", "Electronics", "Clothing",
        "Furniture", "Electronics", "Clothing", "Furniture"
    ],
    "Region": [
        "West", "East", "North", "South",
        "West", "East", "North", "South",
        "West", "East", "North", "South",
        "West", "East", "North", "South",
        "West", "East", "North", "South",
        "West", "East", "North", "South"
    ],
    "Sales": [
        1200, 800, 1500, 2200, 900, 1700,
        1800, 750, 1400, 2500, 1100, 1900,
        1600, 950, 2100, 2800, 1000, 1600,
        2000, 850, 1800, 3000, 1200, 2200
    ],
    "Profit": [
        240, 120, 300, 440, 135, 340,
        360, 110, 280, 500, 165, 380,
        320, 142, 420, 560, 150, 320,
        400, 127, 360, 600, 180, 440
    ],
    "Quantity": [
        4, 5, 3, 6, 7, 4,
        5, 3, 6, 8, 4, 5,
        6, 5, 4, 7, 6, 3,
        5, 4, 7, 9, 5, 6
    ]
}

df = pd.DataFrame(data)
print("First 5 rows:")
print(df.head())
print("\nDataset shape:")
print(df.shape)
print("\nColumn information:")
df.info()
print("\nMissing values:")
print(df.isnull().sum())
```

```python
#Step 3 — Distribution Analysis (Analyze Sales distribution)

plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="Sales", bins=8, kde=True)
plt.title("Distribution of Retail Sales")
plt.xlabel("Sales Amount")
plt.ylabel("Frequency")
plt.show()
```

```python
#Step 4 — Category Analysis using Countplot

plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="Category")
plt.title("Number of Records by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Number of Records")
plt.show()
```

```python
#Step 5 — Outlier Analysis using a Boxplot

plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="Sales")
plt.title("Boxplot of Retail Sales")
plt.xlabel("Sales Amount")
plt.show()
```

```python
#Step 6 — Violin Plot

plt.figure(figsize=(8, 5))
sns.violinplot(data=df, y="Sales", inner="quartile")
plt.title("Violin Plot of Retail Sales")
plt.xlabel("Retail Sales")
plt.ylabel("Sales Amount")
plt.show()
```

```python
#Step 7 — Correlation Heatmap

# Select only numerical columns
numeric_df = df[["Sales", "Profit", "Quantity"]]
# Calculate correlation
correlation_matrix = numeric_df.corr()
# Create the heatmap
plt.figure(figsize=(8, 5))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    fmt=".2f"
)
plt.title("Correlation Heatmap of Retail Data")
plt.show()
```

```python
#Step 8 — Pairplot Analysis

sns.pairplot(
    df,
    vars=["Sales", "Profit", "Quantity"],
    diag_kind="hist"
)
plt.suptitle("Pairplot of Retail Sales Data", y=1.02)
plt.show()
```
