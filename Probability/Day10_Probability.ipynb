```python
#Problem 1: Basic Probability — Rolling a Die
total_outcomes = 6
favorable_outcomes = 3  # Even numbers: 2, 4, 6
probability = favorable_outcomes / total_outcomes
print("Probability of getting an even number:", probability)
print("Probability in percentage:", probability * 100, "%")
```

```python
#Problem 2: Conditional Probability — Customers Using a Coupon
customers_who_purchased = 50
customers_who_used_coupon = 30
probability = customers_who_used_coupon / customers_who_purchased
print("Probability of using a coupon:", probability)
print("Probability in percentage:", probability * 100, "%")
```

```python
#Problem 3: Dependent Events — Selecting Two Red Balls
red_balls = 5
total_balls = 8
# Probability of selecting the first red ball
first_red_probability = red_balls / total_balls

# One red ball has been removed
remaining_red_balls = red_balls - 1
remaining_total_balls = total_balls - 1

# Probability of selecting the second red ball
second_red_probability = remaining_red_balls / remaining_total_balls

# Probability of both events happening
probability = first_red_probability * second_red_probability
print("Probability of selecting two red balls:", probability)
print("Probability in percentage:", probability * 100, "%")
```

```python
#Problem 4: Bayes’ Theorem — Medical Test
disease_probability = 0.01
no_disease_probability = 1 - disease_probability
positive_if_disease = 0.90
positive_if_no_disease = 0.05

# Total probability of receiving a positive test
total_positive_probability = (
    positive_if_disease * disease_probability
    + positive_if_no_disease * no_disease_probability
)

# Probability of having the disease given a positive result
result = (
    positive_if_disease * disease_probability
    / total_positive_probability
)

print("Probability of disease after a positive test:", round(result, 4))
print("Probability in percentage:", round(result * 100, 2), "%")
```

```python
#Problem 5: Binomial Distribution — Marketing Email Clicks
import math

n = 5               # Total customers
k = 2               # Required number of clicks
p = 0.20            # Probability of a click
q = 1 - p           # Probability of no click

# Calculate the binomial probability
combinations = math.comb(n, k)
probability = combinations * (p ** k) * (q ** (n - k))

print("Probability of exactly two clicks:", probability)
print("Probability in percentage:", probability * 100, "%")
```
