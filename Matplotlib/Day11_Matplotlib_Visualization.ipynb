```python
#Step 1: Import libraries and create our dataset

import pandas as pd
import matplotlib.pyplot as plt
# Create a sample retail sales dataset
data = {
    "Month": [
        "Jan", "Feb", "Mar", "Apr", "May", "Jun",
        "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "Sales": [
        10000, 12000, 14000, 16000, 18000, 17000,
       19000, 22000, 20000, 23000, 21000, 18000],
    "Profit": [
        2000, 2500, 3000, 3500, 4000, 3800,
        4200, 5000, 4500, 5500, 4800, 4000],
    "Quantity": [
        50, 60, 70, 80, 90, 85,
        95, 110, 100, 115, 105, 90]
}
df = pd.DataFrame(data)
# Display the dataset
print(df)
```

```python
#Step 2: Create a line chart

plt.figure(figsize=(10, 5))
plt.plot(df["Month"], df["Sales"], marker="o", label="Sales")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales (₹)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Step 3: Create a bar chart

plt.figure(figsize=(10, 5))
plt.bar(df["Month"], df["Sales"], label="Sales")
plt.title("Monthly Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales (₹)")
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```

```python
#Step 4: Create a histogram

plt.figure(figsize=(10, 5))
plt.hist(df["Sales"], bins=5, edgecolor="black")
plt.title("Distribution of Monthly Sales")
plt.xlabel("Sales (₹)")
plt.ylabel("Frequency")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```

```python
#Step 5: Create a scatter plot

plt.figure(figsize=(10, 5))
plt.scatter(df["Quantity"], df["Sales"], color="blue", label="Monthly Sales")
plt.title("Relationship Between Quantity and Sales")
plt.xlabel("Quantity Sold")
plt.ylabel("Sales (₹)")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

```python
#Step 6: Create a pie chart

plt.figure(figsize=(8, 8))
plt.pie(
    df["Sales"],
    labels=df["Month"],
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Monthly Contribution to Total Sales")
plt.show()
```

```python
#Step 7: Create a box plot

plt.figure(figsize=(8, 5))
plt.boxplot(df["Sales"])
plt.title("Distribution of Monthly Sales")
plt.ylabel("Sales (₹)")
plt.xticks([1], ["Monthly Sales"])
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```

```python
#Step 8: Create an area chart

plt.figure(figsize=(10, 5))
plt.fill_between(df["Month"], df["Sales"], alpha=0.4, label="Sales")
plt.plot(df["Month"], df["Sales"], marker="o")
plt.title("Monthly Sales Area Chart")
plt.xlabel("Month")
plt.ylabel("Sales (₹)")
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```

```python
#Step 9: Create a profit line chart

plt.figure(figsize=(10, 5))
plt.plot(
    df["Month"],
    df["Profit"],
    marker="o",
    color="green",
    label="Profit"
)
plt.title("Monthly Profit Trend")
plt.xlabel("Month")
plt.ylabel("Profit (₹)")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

```python
#Step 10: Create a quantity bar chart

plt.figure(figsize=(10, 5))
plt.bar(df["Month"], df["Quantity"], color="orange", label="Quantity Sold")
plt.title("Monthly Quantity Sold")
plt.xlabel("Month")
plt.ylabel("Quantity Sold (Units)")
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```

```python
#Step 11: Create the final histogram

plt.figure(figsize=(10, 5))
plt.hist(
    df["Sales"],
    bins=10,
    color="skyblue",
    edgecolor="black"
)
plt.title("Detailed Distribution of Monthly Sales")
plt.xlabel("Sales (₹)")
plt.ylabel("Number of Months")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```
