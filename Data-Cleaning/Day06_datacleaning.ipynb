```python
# Step 1 - Import pandas

import pandas as pd
```

```python
# Step2 - Load the dirty dataset

df = pd.read_csv("dirty_dataset.csv")
print(df)
```

```python
# Step 3 - Look at the dataset BEFORE cleaning

print("Dataset Before Cleaning:")
print(df)
```

```python
# Step 4 - check its size:
print("Shape:", df.shape)
```

```python
# Step 5 - Keep a backup of the original data

before_cleaning = df.copy()
```

```python
# Step 6 - Identify missing values

print(df.isnull().sum())
```

```python
# Step 7 - Count duplicate rows

print("Number of duplicate rows:", df.duplicated().sum())
```

```python
# Step 8 - Check current data types

print(df.dtypes)
```

```python
# Step 9 - Fix the "signup date" column

df["signup date"] = pd.to_datetime(
    df["signup date"],
    errors="coerce"
)
```

```python
# Step 10 - Check the result
print(df["signup date"])
```

```python
# Step 11 - Standardize text
df["Full Name"] = df["Full Name"].str.strip()
df["country"] = df["country"].str.strip()
df["is_active"] = df["is_active"].str.strip()
```

```python
# Step 12 - Check the output for step11
print(df[["Full Name", "country", "is_active"]])
```

```python
# Step 13 - Standardize country values

df["country"] = df["country"].replace({
    "U.S.A.": "USA",
    "United States": "USA",
    "china": "China",
    "DE": "Germany",
    "United Kingdom": "UK"
})
print(df["country"])
```

```python
# Step 14 - Standardize the is_active column

df["is_active"] = df["is_active"].replace({
    "yes": "Yes",
    "Yes": "Yes",
    "TRUE": "Yes",
    "true": "Yes",
    "1": "Yes",
    "Y": "Yes",
    "active": "Yes",
    "No": "No",
    "no": "No",
    "FALSE": "No",
    "0": "No"
})
print(df["is_active"])
```

```python
# Step 15 - Standardize names
df["Full Name"] = df["Full Name"].str.title()
print(df["Full Name"])
```

```python
# Step 16 - Handle duplicates
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
```

```python
# Step 17 - check output for the step16
print("Duplicates after cleaning:", df.duplicated().sum())
```

```python
# Step 18 - Clean the amount column
df["amount"] = (
    df["amount"]
    .str.replace("$", "", regex=False)
    .str.replace("€", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)
print(df["amount"])
```

```python
# Step 19 - Convert amount into numbers
df["amount"] = pd.to_numeric(
    df["amount"],
    errors="coerce"
)
print(df["amount"])
```

```python
# Step 20 - Handle missing values
print(df.isnull().sum())
df = df.dropna()
#This shows us the rows containing missing values.
print(df[df.isnull().any(axis=1)])
```

```python
# Step 21 - Identify abnormal values
print(df["amount"].sort_values())
```

```python
print(df[df["amount"] < 0])
```

```python
#Step 18 — Check the cleaned dataset
print("Cleaned Dataset:")
print(df)
```

```python
# Checking final dataset shape
print("Final Shape:", df.shape)
```

```python
# Checking data types of the final dataset 
print(df.dtypes)
```

```python
print(df.isnull().sum()) #This lets us confirm that our cleaning steps worked.
```

```python
# Step 19 - Save the cleaned data set
df.to_csv("cleaned_dataset.csv", index=False)
```
