```python
import pandas as pd
import numpy as np

# Create a sample retail sales dataset
data = {
    "Sales": [1000, 1500, 2000, 2000, 2500, 3000, 3500, 4000, 4500, 10000],
    "Quantity_Sold": [10, 12, 15, 16, 18, 20, 22, 25, 28, 50]
}
# Convert the data into a DataFrame
df = pd.DataFrame(data)
# Display the dataset
print(df)
```

```python
# Calculate mean
print("Mean Sales:", df["Sales"].mean())
```

```python
# Calculate median
print("Median Sales:", df["Sales"].median())
```

```python
# Calculate mode
print("Mode Sales:", df["Sales"].mode().tolist())
```

```python
# Calculate range
sales_range = df["Sales"].max() - df["Sales"].min()
print("Range:", sales_range)
```

```python
# Calculate population variance
sales_variance = df["Sales"].var(ddof=0)
print("Population Variance:", sales_variance)
```

```python
# Calculate population standard deviation
sales_std = df["Sales"].std(ddof=0)
print("Population Standard Deviation:", round(sales_std, 2))
```

```python
# Calculate percentiles
p25 = np.percentile(df["Sales"], 25)
p50 = np.percentile(df["Sales"], 50)
p75 = np.percentile(df["Sales"], 75)
p90 = np.percentile(df["Sales"], 90)
print("25th Percentile:", p25)
print("50th Percentile:", p50)
print("75th Percentile:", p75)
print("90th Percentile:", p90)
```

```python
# Calculate quartiles
q1 = df["Sales"].quantile(0.25)
q2 = df["Sales"].quantile(0.50)
q3 = df["Sales"].quantile(0.75)
print("First Quartile (Q1):", q1)
print("Second Quartile (Q2):", q2)
print("Third Quartile (Q3):", q3)
```

```python
# Calculate Interquartile Range
iqr = q3 - q1
print("Interquartile Range (IQR):", iqr)
```

```python
# Calculate correlation between Sales and Quantity Sold
correlation = df["Sales"].corr(df["Quantity_Sold"])
print("Correlation:", round(correlation, 4))
```

```python
# Calculate covariance between Sales and Quantity Sold
covariance = df["Sales"].cov(df["Quantity_Sold"])
print("Covariance:", round(covariance, 2))
```
