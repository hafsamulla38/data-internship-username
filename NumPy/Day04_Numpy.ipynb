```python
# Day-4 Task
# Step 1 : Import NumPy and create sales data

import numpy as np
sales = np.array([100, 200, 150, 300, 250, 350, 400, 275, 225, 500])
print("Sales Data:")
print(sales)
```

```python
# Step 2 : Total Sales

total_sales = np.sum(sales)
print("Total Sales:", total_sales)
```

```python
# Step 3 : Average Sales
average_sales = np.mean(sales)
print("Average Sales:", average_sales)
```

```python
# Step 4 : Maximum Sales
maximum_sales = np.max(sales)
print("Maximum Sales:", maximum_sales)
```

```python
# Step 5 : Minimum Sales
minimum_sales = np.min(sales)
print("Minimum Sales:", minimum_sales)
```

```python
# Step 6 : Standard Deviation
standard_deviation = np.std(sales)
print("Standard Deviation:", standard_deviation)
```

```python
# Step 7 : Median
median_sales = np.median(sales)
print("Median Sales:", median_sales)
```

```python
# Step 8 : 25th Percentile
percentile_25 = np.percentile(sales, 25)
print("25th Percentile:", percentile_25)
```

```python
# Step 9 : 75th Percentile
percentile_75 = np.percentile(sales, 75)
print("75th Percentile:", percentile_75)
```

```python
# Step 10 : Practice of 1D array

sales = np.array([100, 200, 150, 300, 250])
print(sales)
print("Dimensions:", sales.ndim)
```

```python
# Step 11 : Practice 2D Array

sales_2d = np.array([
    [100, 200, 150],
    [300, 250, 400]
])
print(sales_2d)
```

```python
# Step 12 : Checking its dimension and shape
print("Dimensions:", sales_2d.ndim)
print("Shape:", sales_2d.shape)
```

```python
# Step 13 : Practicing Reshaping
numbers = np.array([1, 2, 3, 4, 5, 6])
new_array = numbers.reshape(2, 3)
print(new_array)
```

```python
# Step 13 : Matrix Multiplication
matrix1 = np.array([
    [1, 2],
    [3, 4]
])
matrix2 = np.array([
    [5, 6],
    [7, 8]
])
result = np.matmul(matrix1, matrix2)
print(result)
```
